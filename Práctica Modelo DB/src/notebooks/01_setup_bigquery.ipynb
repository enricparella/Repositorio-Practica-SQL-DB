# Setup de BigQuery

En este notebook se configura la conexión con Google Cloud BigQuery y se crean el dataset y las tablas necesarias para el modelo de datos de KTE E-commerce.

```python
import os
from dotenv import load_dotenv
from google.cloud import bigquery

load_dotenv()

PROJECT_ID = os.getenv("GCP_PROJECT_ID")
DATASET_ID = os.getenv("BQ_DATASET_ID")
```

```python
client = bigquery.Client(project=PROJECT_ID)

print(f"Connectado al proyecto: {client.project}")
print(f"Dataset: {DATASET_ID}")
```

```python
dataset_ref = bigquery.Dataset(f"{PROJECT_ID}.{DATASET_ID}")
dataset_ref.location = "EU"

dataset = client.create_dataset(dataset_ref, exists_ok=True)

print(f"Dataset preparado: {dataset.dataset_id}")
```

## Creación de tablas

A continuación se prepararán todas las tablas necesarias para el proyecto de la BD.

### Tabla customers

```python
schema_customers = [
    bigquery.SchemaField("customer_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("first_name", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("last_name", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("email", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("phone", "STRING"),
    bigquery.SchemaField("country", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("city", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("acquisition_channel", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("registration_date", "DATE", mode="REQUIRED"),
]

table_id = f"{PROJECT_ID}.{DATASET_ID}.customers"

table = bigquery.Table(table_id, schema=schema_customers)
table = client.create_table(table, exists_ok=True)

print(f"Tabla creada: {table.table_id}")
```

### Tabla categorías

```python
schema_categories = [
    bigquery.SchemaField("category_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("name", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("description", "STRING"),
]

table_id = f"{PROJECT_ID}.{DATASET_ID}.categories"

table = bigquery.Table(table_id, schema=schema_categories)
table = client.create_table(table, exists_ok=True)

print(f"Tabla creada: {table.table_id}")
```

### Tabla productos

```python
schema_products = [
    bigquery.SchemaField("product_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("category_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("name", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("description", "STRING"),
    bigquery.SchemaField("sale_price", "FLOAT", mode="REQUIRED"),
    bigquery.SchemaField("cost_price", "FLOAT", mode="REQUIRED"),
    bigquery.SchemaField("stock", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("is_active", "BOOLEAN", mode="REQUIRED"),
]

table_id = f"{PROJECT_ID}.{DATASET_ID}.products"

table = bigquery.Table(table_id, schema=schema_products)
table = client.create_table(table, exists_ok=True)

print(f"Tabla creada: {table.table_id}")
```

### Tabla pedidos

```python
schema_orders = [
    bigquery.SchemaField("order_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("customer_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("status", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("shipping_address", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("shipping_city", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("shipping_country", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("order_date", "DATE", mode="REQUIRED"),
    bigquery.SchemaField("shipping_date", "DATE"),
    bigquery.SchemaField("delivery_date", "DATE"),
]

table_id = f"{PROJECT_ID}.{DATASET_ID}.orders"

table = bigquery.Table(table_id, schema=schema_orders)
table = client.create_table(table, exists_ok=True)

print(f"Tabla creada: {table.table_id}")
```

### Tabla líneas de pedido

```python
schema_order_items = [
    bigquery.SchemaField("order_item_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("order_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("product_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("quantity", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("unit_price", "FLOAT", mode="REQUIRED"),
    bigquery.SchemaField("discount", "FLOAT"),
]

table_id = f"{PROJECT_ID}.{DATASET_ID}.order_items"

table = bigquery.Table(table_id, schema=schema_order_items)
table = client.create_table(table, exists_ok=True)

print(f"Tabla creada: {table.table_id}")
```

### Tabla pagos

```python
schema_payments = [
    bigquery.SchemaField("payment_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("order_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("payment_method", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("status", "STRING", mode="REQUIRED"),
    bigquery.SchemaField("amount", "FLOAT", mode="REQUIRED"),
    bigquery.SchemaField("payment_date", "DATE", mode="REQUIRED"),
]

table_id = f"{PROJECT_ID}.{DATASET_ID}.payments"

table = bigquery.Table(table_id, schema=schema_payments)
table = client.create_table(table, exists_ok=True)

print(f"Tabla creada: {table.table_id}")
```

### Tabla valoraciones

```python
schema_reviews = [
    bigquery.SchemaField("review_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("order_item_id", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("rating", "INTEGER", mode="REQUIRED"),
    bigquery.SchemaField("comment", "STRING"),
    bigquery.SchemaField("review_date", "DATE", mode="REQUIRED"),
]

table_id = f"{PROJECT_ID}.{DATASET_ID}.reviews"

table = bigquery.Table(table_id, schema=schema_reviews)
table = client.create_table(table, exists_ok=True)

print(f"Tabla creada: {table.table_id}")
```

## Verificación de tablas creadas

```python
tables = list(client.list_tables(f"{PROJECT_ID}.{DATASET_ID}"))

print("Tablas disponibles:")

for table in tables:
    print(f"- {table.table_id}")
```
